# 01 Load and Clean AIID Data

This notebook is a lightweight companion to the scripted pipeline. It previews the raw AIID extract and the cleaned incident table built by `src/load_aiid_data.py` and `src/clean_incidents.py`.

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
raw_path = ROOT / 'data' / 'raw' / 'aiid_snapshot_raw.csv'
clean_path = ROOT / 'data' / 'processed' / 'incidents_cleaned.csv'

raw_df = pd.read_csv(raw_path)
clean_df = pd.read_csv(clean_path)

print('Raw rows:', len(raw_df))
print('Cleaned rows:', len(clean_df))
print('Cleaned year range:', clean_df['incident_year'].min(), 'to', clean_df['incident_year'].max())
```
